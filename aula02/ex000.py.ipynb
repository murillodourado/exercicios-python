Exemplo 1 - Cálculo da Média de um Aluno

```python
# Cálculo da média de um aluno

# Variáveis
n1 = 9
n2 = 10
n3 = 7

# Cálculo da média
media_aluno = (n1 + n2 + n3) / 3 

# Exibir os resultados
print('Nota 1 =', n1)
print('Nota 2 =', n2)
print('Nota 3 =', n3)
print(f'Média = {media_aluno:.1f}')
```

Exemplo 2 - Cálculo da Média de um aluno com input para o usuário

```python
# Variáveis para o usuário 
nota1 = float(input('Insira nota 1: '))
nota2 = float(input('Insira nota 2: '))
nota3 = float(input('Insira nota 3: '))

# Cálculo da média
media_aluno = (nota1 + nota2 + nota3) / 3 

# Lógica para aprovação ou não do aluno
if media_aluno > 7:
    print('Aprovado!')
else:
    print('Reprovado!')

# print(nota1)
# print(nota2)
# print(nota3)
# print(media_aluno)
```

Exercício 1. Crie um algoritmo para calcular a área de um triângulo, área de um círculo e área de um retângulo. Crie uma estrutura de input para o usuário e mostre os valores das áreas de cada figura geométrica.

```python
# Aréa de um triângulo

valor_base = int(input('Digite o valor da base do triângulo (em cm): '))
valor_altura = int(input('Digite o valor da altura do triângulo (em cm): '))
area_triangulo = (valor_base * valor_altura) / 2

print(f'O valor da área de um triângulo com base {valor_base} e altura {valor_altura} é {area_triangulo:.0f} cm')
```

```python
# Área de um círculo

valor_pi = 3.14
valor_raio = float(input('Digite o valor do raio do círculo (em cm): '))
area_circulo = valor_pi * (valor_raio ** 2)

print(f'O valor da área do círculo é {area_circulo} cm')
```

```python
# Aréa de um retângulo 

valor_base = float(input('Digite o valor da base (em cm): '))
valor_altura = float(input('Digite o valor da altura (em cm): '))
area_retangulo = valor_base * valor_altura

print(f'O valor da área do retângulo é {area_retangulo:.0f} cm')
```
